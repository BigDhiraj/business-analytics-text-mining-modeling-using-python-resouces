```python
# write a function to expand contractions in the text

def expand_contractions(sent, mapping):

# pattern for matching contractions with their expansions
    pattern = re.compile('({})'.format('|'.join(mapping.keys())), flags=re. IGNORECASE|re.DOTALL)

    def expand_map(contraction):
        # using group method to access subgroups of the match
        match = contraction.group(0)

        # to retain the correct case of the word
        first_char = match[0]

        # find out the expansion
        expansion = mapping.get(match) if mapping.get(match) else mapping.get(match.lower())
        
        expansion = first_char + expansion[1:]
        return expansion

    # using sub method to replace all contractions with their expansions for a sentence
    # function expand map will be called for every non-overlapping occurrence of the pattern
    expand_sent = pattern.sub(expand_map, sent)
    return expand_sent
```

```python
## correcting words ##
# example: spelling mistakes, repeated characters

# repeated characters

# using a regex pattern and match groups (groups 1, 2, and 3) with the pattern r'\1\2\3'
# using wordnet corpus to check for valid words

# write a function to eliminate repeated characters from tokens
from nltk.corpus import wordnet

def remove_repeated_characters(tokens):
    # characters that occur twice among other characters
    pattern = re.compile(r'(\w*)(\w)\2(\w*)')
    
    # using backreferences to construct the replacement for the pattern
    # e.g., \1 will be replaced with the substring matched by group 1 in the pattern
    # keep one insatnce of the repeating character in group 2
    repl = r'\1\2\3'
    
    def replace(tkn):
        # check for semantically correct word
        if wordnet.synsets(tkn):
            return tkn
        # replace the pattern
        tkn_c = pattern.sub(repl, tkn)
        
        # recursive call
        return replace(tkn_c) if tkn_c != tkn else tkn_c
        
    # correct each token
    tokens_c = [replace(tn) for tn in tokens]
    return tokens_c
```

```python
## write a function to combine or chain all the transformation steps for a text corpus
#includes expanding contractions, PoS tagging, Lemmatization,
#special character removal, stopword removal, tokenization

import re
import nltk
import string

# Load all the English stopwords
seq_sw1 = nltk.corpus.stopwords.words('english')

# Load contraction mappings
# contraction_map

# Load the Lemmatizer
from nltk.stem import WordNetLemmatizer
wnl1 = WordNetLemmatizer()

# define a function to tokenize text into tokens
def tokenize_text(text):
    tokens = nltk.word_tokenize(text)
    tokens = [tn.strip() for tn in tokens]
    return tokens

# define a function for expanding contractions
def expand_contractions(text, mapping):

    # pattern for matching contractions with their expansions
    pattern = re.compile('({})'.format('|'.join(mapping.keys())), flags=re. IGNORECASE|re.DOTALL)
    
    def expand_map(contraction):
        # using group method to access subgroups of the match
        match = contraction.group(0)

        # to retain the correct case of the word
        first_char = match[0]
        
        # find out the expansion
        expansion = mapping.get(match) if mapping.get(match) else mapping.get(match.lower())
        expansion = first_char + expansion[1:]
        return expansion
        
    # using sub method to replace all contractions with their expansions for a sentence
    # function expand_map will be called for every non-overlapping occurrence of the pattern
    expand_text = pattern.sub(expand_map, text)
    return expand_text


# define a function for lemmatization of text based on POS tags

from nltk.corpus import wordnet as wn

# Annotate text tokens with POS tags
def pos_tag_text(text):

    # convert Penn treebank tag to wordnet tag
    # since the WordNetLemmatizer checks for POS tag annotations based on WordNet formats
    def penn_to_wn_tags(pos_tag):
        if pos_tag.startswith(')'):
            return wn.ADJ
        elif pos_tag.startswith('V'):
            return wn. VERB
        elif pos_tag.startswith('N'):
            return wn. NOUN
        elif pos_tag.startswith('R'):
            return wn.ADV
        else:
            return None
    
    # tokenize text into tokens using tokenize_text function
    tokens = tokenize_text(text)
    
    # using pos_tag function to annotate POS tags
    tokens_t = nltk.pos_tag(tokens, tagset='universal')

    # convert the POS tags from the Penn treebank syntax to WordNet syntax
    tokens_p = [(word.lower(), penn_to_wn_tags(pos_tag)) for word, pos_tag in tokens_t]
    
    return tokens_p

# Lemmatize text based on POS tags
def lemmatize_text(text):

    # PoS tagging
    tokens_p = pos_tag_text(text)
    
    # Lemmatize the text
    tokens_lm = [wnl1. lemmatize(word, pos_tag) if pos_tag else word for word, pos_tag in tokens_p]
    
    # concatenate tokens
    text_lm = ''.join(tokens_lm)
    
    return text_lm

# define a function to remove special symbols and characters
def remove_special_characters(text):

    # tokenize text into tokens using tokenize_text function
    tokens = tokenize_text(text)

    pattern = re.compile('[{}]' .format(re.escape(string.punctuation)))
    tokens_f = filter(None, [pattern.sub('', tn) for tn in tokens])
    
    text_f = ''.join(tokens_f)
    
    return text_f

# define a function to remove stopwords from our text data
def remove_stopwords(text):

    # tokenize text into tokens using tokenize_text function
    tokens = tokenize_text(text)
    
    # using English stopwords list for removal
    tokens_f = [tn for tn in tokens if tn not in seq_sw1]
    
    text_f = ' '.join(tokens_f)
    
    return text_f

# define a function to chain all the transformation steps
def transform_corpus(corpus, tokenize=False):

    corpus_t = []

    for text in corpus:
        text = expand_contractions(text, contraction_map)
        text = lemmatize_text(text)
        text = remove_special_characters(text)
        text = remove_stopwords(text)
        
        corpus_t.append(text)
        
        if tokenize:
            text = tokenize_text(text)
            corpus_t.append(text)

    return corpus_t
```

```python
## bag of words model
# simplest vectorization model

# define a function to create a bag of words feature extraction or vectorization model

from sklearn.feature_extraction. text import CountVectorizer

def bow_extractor(corpus, ngram_range=(1,1)):

    # using an instance of CountVectorizer class to define the type of features to be extracted
    # first argument: terms having a minimum frequency of 1 (default)
    # second argument: default unigrams only: 1 means uni, 2 means bi, 3 means tri and
    #                  range of (1,3) would mean uni to tri (uni, bi, tri)
    vectorizer = CountVectorizer(min_df=1, ngram_range=ngram_range)
    
    # using fit_transform method to extarct the features from corpus (term-document matrix)
    features = vectorizer.fit_transform(corpus)
    
    return vectorizer, features
    
    #
```

```python
## TF-IDF model
# solves the potential problem of bow model:
# words with higher frequencies across documents might dominate
# more relevant words (w.r.t classes to be predicted) with Lower frequencies

# define a function to create a tf-idf feature extraction or vectorization model based on bow features

from sklearn.feature_extraction.text import TfidfTransformer

def tfidf_transformer(bow_features):

    # using an instance of TfidfTransformer class to define the type of features to be extracted
    # first argument: Euclidean normalization (default: 'L2')
    # second argument: adding one to document frequencies to prevent zero divisions
    transformer = TfidfTransformer(norm='l2', smooth_idf=True, use_idf=True)
    
    # using fit_transform method to extract the features from bow features (term-document matrix)
    features = transformer.fit_transform(bow_features)
    
    return transformer, features
```

```python
# define a function to remove empty text documents

def remove_empty_docs(corpus, labels):
    
    corpus_f = []
    labels_f = []
    
    for doc, lbl in zip(corpus, labels):
        if doc.strip():
            corpus_f.append(doc)
            labels_f.append(lbl)
    
    return corpus_f, labels_f
```

```python
# evaluation metrics

# define a function to compute evaluation metrics for classification models

from sklearn import metrics
import numpy as np

def get_metrics(actual_labels, predicted_labels):

    print('Accuracy:', np.round(metrics.accuracy_score(actual_labels, predicted_labels), 2))
    
    print('Precision:', np.round(metrics.precision_score(actual_labels, predicted_labels, average='weighted'), 2))
    
    print('Recall:', np.round(metrics.recall_score(actual_labels, predicted_labels, average='weighted'), 2))
    
    print('F1 Score:', np.round(metrics.f1_score(actual_labels, predicted_labels, average='weighted'), 2))
```

```python
# define a function to build the model using training dataset, then evaluates the model performance on test dataset

def train_predict_evaluate_model(classifier, train_f, train_1, test_f, test_1):
    
    # build model
    classifier.fit(train_f, train_1)
    
    # predict using model
    predictions = classifier.predict(test_f)
    
    # evaluate model prediction performance
    get_metrics(actual_labels=test_1, predicted_labels=predictions)
    
    return predictions
```

```python
# Building a classifier using support vector machine (SVM)

from sklearn. linear_model import SGDClassifier

svm = SGDClassifier(loss='hinge', max_iter=1000, tol=0.2)
```

```python
# Support Vector Machine with bag of words features
svm_bow_predictions = train_predict_evaluate_model(classifier=svm,
                                                   train_f=features_tr1,
                                                   train_1=train_1,
                                                   test_f=features_te,
                                                   test_1=test_1)
```

```python
# Support Vector Machine with tfidf features
svm_tfidf_predictions = train_predict_evaluate_model(classifier=svm,
                                                     train_f=features_tr1,
                                                     train_1=train_1,
                                                     test_f=features_te1,
                                                     test_1=test_1)
```
