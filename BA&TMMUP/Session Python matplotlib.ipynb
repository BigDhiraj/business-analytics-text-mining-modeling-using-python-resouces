```python
# Load required Library modules
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Example: a simple plot
arr1d = np.arange(10)
arr1d
```

```python
# create the plot using plot function
plt.plot(arr1d)
```

```python
## Figure object:
# plots are created in the frame defined by this object

# creating a figure object using figure function
fig = plt.figure()
```

```python
# add axes to the figure object

# using add_subplot method
# first and second arguments are used to specify the grid (no. of rows and columns in the figure)

# e.g., 2 x 2 (four plots)

# thrid argument is to specify the index position in the grid

axes1 = fig.add_subplot(2, 2, 1)
```

```python
axes2 = fig.add_subplot(2, 2, 2)
```

```python
axes3 = fig.add_subplot(2, 2, 3)
```

```python
fig
```

```python
# by default, last subplot is used for plotting

# string abbreviation to indicate color and line style
# e.g., k is for black color and -- is for dashed Line (refer help section)

plt.plot(np.random.randn(50).cumsum(), 'k--')
```

```python
# using empty subplots
# e.g., histogram using hist method
_ = axes1.hist(np.random.randn(100), bins=20, color='k', alpha=0.3)
```

```python
# e.g., scatter plot using scatter method
axes2.scatter(np.arange(30), np.arange(30) + 3 * np.random.randn(30))
```

```python
# creating a figure with a grid of subplots
# using subplots function
# returns an array of subplot objects
fig1, axes = plt.subplots(2, 3)
```

```python
# indexing is like 2darray
axes[0,1].scatter(np.arange(30), np.arange(30) + 3 * np.random.randn(30))
```
