```python
# setdefault
# Useful to set values while generating a dict
# Example: categorizing a list of terms by their first letters as a dict of lists
terms = ['alpha', 'beta', 'gamma', 'alpha2', 'beta2', 'beta3', 'll' ]
d3 = {}
for x in terms:
    d3.setdefault(x[0], []).append(x)
d3
```

```python
# defaultdict class from collection module
# Can also be used to set values while generating a dict
from collections import defaultdict
first_letter = defaultdict(list)
for x in terms:
    first_letter[x[0]].append(x)
first_letter
```

```python
# writing more reusable and generic functions
# exampple: data cleaning on a list of strings, e.g., stripping whitespace ,  removing punctuation
# on proper capitalization

# first approach
# using built-in string methods along with the 're' standard library module for regular expression
messy_strings = [' Alabama ','Georgia!','Georgia','georgia','FlOrIda','south carolina##','West virginia?']
import re
def clean_strings(strings):
    result = []
    for x in strings:
        x = x.strip()
        x = re.sub('[!#?]', '', x)
        x = x.title()
        result.append(x)
    return result
```

```python
messy_strings
```

```python
clean_strings(messy_strings)
```

```python
# second approach
# more generic , organize , usable, scalabel
# applying a list of the operationon a list of strings
def remove_punctuation(value):
    return re.sub('[!#?]', '', value)

# list of function based operations to be applied: Both user-defined and built-in
clean_ops = [str.strip, remove_punctuation, str.title]

def clean_strings(strings, ops):
    result = []
    for x in strings:
        for f in ops:
            x = f(x)
        result.append(x)
    return result
```

```python
messy_strings
```

```python
clean_strings(messy_strings, clean_ops)
```

```python
# functions as arguments to other functions
# example: 'map' function
# applies a function to a sequence
for x in map(remove_punctuation, messy_strings):
    print(x)
```

```python
# Generators return a sequence of multiple elements in an orderly fashion as and when requested
# generators use the 'yield' keyword instead of 'return'
# Example:
def squares(n = 10):
    print('Generating squares from 1 to {0}'.format(n ** 2))
    for i in range(1, n + 1):
        yield i ** 2
```

```python
## Example: random walks
# a simple random walk starting at 0 with steps of 1 and -1 occurring with equal probability

# using Python built-in capabilities
import random
x = 0
walk = [x]
nsteps = 1000
for i in range(nsteps):
    step = 1 if random.randint(0, 1) else -1
    x += step
    walk.append(x)
```

```python
# multiple random walks
# e.g., 50 walks of 1000 steps each
nwalks = 50
nsteps = 1000
draws1 = np.random.randint(0, 2, size=(nwalks, nsteps)) # 0 or 1
steps2 = np.where(draws1 > 0, 1, -1)
walks2 = steps2.cumsum(1) # along second dimension
```

```python
steps2
```

```python
walks2
```

```python
walks2.shape
```

```python
walks2.min()
```

```python
walks2.max()
```

```python
# minimum crossing time for any of the walks to reach a particular value
# e.g., absolute '30' level
# first find out which of the walks reach the value
hits = (np.abs(walks2) >= 30).any(1) # along second dimension
```

```python
hits
```

```python
# no. of walks hitting '30' (out of 50)
hits.sum()
```

```python
firsthits = (np.abs(walks2[hits]) >= 30).argmax(1) # along second dimension
```

```python
firsthits
```

```python
mct = firsthits.min()
mct
```

```python
mct_ind = list(firsthits == mct).index(True)
mct_ind
```

```python
# Index of the True value which is 'mct_ind + 1'th among True values in hits array
count_true = 0
for i in range(len(hits)):
    if count_true <= mct_ind:
        if hits[i] == True:
            count_true += 1
        else:
            continue
    else:
        break
```
