```python
import yfinance as yf

dataset = {
    ticker: yf.download(
        ticker,
        start='2010-01-01',
        end='2020-01-01',
        auto_adjust=False,
        multi_level_index=False
    )
    for ticker in ['AAPL', 'IBM', 'MSFT', 'GOOG']
}
```

```python
price = pd.DataFrame({ticker: data['Adj Close'] for ticker, data in dataset.items()})# not worked in newer version
```

```python
price = pd.DataFrame({
    ticker: data['Adj Close']
    for ticker, data in dataset.items()
})
```

```python
price.head()
```

```python
volume = pd.DataFrame({ticker: data['Volume'] for ticker, data in dataset.items()})
```

```python
volume.head()
```

```python
# percent changes of the prices
returns = price.pct_change()
```

```python
# Last (most recent) 5 cases
returns.tail()
```

```python
# compute the correlations
# using corr method
# between two columns
returns['MSFT'].corr(returns[ 'IBM' ])
```

```python
returns['MSFT'].cov(returns[ 'IBM' ])
```

```python
# using column name as attribute
returns.MSFT.corr(returns. IBM)
```

```python
# using column name as attribute
returns.MSFT.corr(returns. IBM)
```

```python
# correlation matrix
returns.corr()
```

```python
# compute the covariances
# covariance matrix
returns.cov()
```

```python
# pairwise correlations with one particular column
# using corrwith method
returns.corrwith(returns. IBM)
```

```python
# correlations between columns of different dataframes having same column names
returns.corrwith(volume)
```
